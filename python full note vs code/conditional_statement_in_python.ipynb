### Conditional Statement (if elif else) ###

A condition statement in Python is a statement that checks a condition and decides which code to run based on whether the condition is True or False.






### 1. 'if' conditional statement



An if statement in Python is used to check a condition. If the condition is True, Python executes the code inside the if block.

```python
a = 6
b = 10
if b>a:   #works only for true condition
  print("b is greater than a")
```

```python
age = 20

if age > 18:
    print("You are an adult")
```

```python
age = int(input("enter your age: "))
if age>19:
    print("you're an adult")
```

### 2. if-else Statement

An if-else statement is used when a program has two choices.

If the condition is True → if block runs.

If the condition is False → else block runs.

```python
temperature = 30
if temperature < 30: #condition false-else run
  print("it's a hot day")
else:
  print("it's a cold day")
```

```python
age = int(input("enter your age: "))
if age > 20:
  print("you're an adult")
else:
  print("you're not an adult") #else handles false condition
```

### 3. if-elif-else Statement

An if-elif-else statement is used when a program has more than two choices.

if → checks the first condition.

elif → checks another condition if the first is false.

else → runs when all conditions are false.

```python
score = 85
if score>= 90:
  print("grade-a")
elif score>= 80:
  print("grade-b")
elif score>=70:
  print("grade-c")
else:
  print("grade-d")
```

### 4. Nested if-else — Definition

A nested if-else means putting one if-else statement inside another if or else statement.

It is used when we need to check a second condition after checking the first condition.

```python
number = int(input("enter your number"))
if number > 0:
  if number % 2 == 0:
    print("The number is positive and even. ")
  else:
    print("The number is positive and odd. ")
else:
  if number == 0:
    print("The number is zero. ")
  else:
    print("The number is negative. ")
```

### 5. Conditional Statement

A conditional statement is used in programming to make a decision based on a condition.

If the condition is True, one action happens. If it is False, another action may happen.

```python
age = int(input("enter your age"))
status = "adult" if age>=20 else "minor"
print(status)
```

```python
age = 18
status = "adult" if age>=20 else "minor"
print(status)
```

```python
value = None
if value:
  print("value is true")
else:
  print("value is false")
```

```python
year = int(input("enter a year (e.g. 2024): "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is leap year")
else:
    print(f"{year} is not leap year")
```

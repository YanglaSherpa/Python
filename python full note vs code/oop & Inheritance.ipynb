### OOP (Object-Oriented Programming)

OOP stands for Object-Oriented Programming. It is a way of writing programs by organizing code around objects that contain data (attributes) and functions (methods).

Example

Car:

Attributes (data): brand, color, model, speed

Methods (actions): start(), stop(), accelerate(), brake()

OOP is useful because it makes programs:

Easy to organize – code is divided into classes and objects.

Reusable – we can create many objects from one class.

Easy to maintain – changes can be made in one place.

Secure – data can be controlled using encapsulation.

Suitable for large programs – complex systems can be divided into smaller parts.

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def start(self):
        print("Car is starting")

    def stop(self):
        print("Car is stopping")


car1 = Car("Toyota", "Red")

print(car1.brand)
print(car1.color)

car1.start()
car1.stop()
```

Class

A class is a blueprint for creating objects.

Example: A Student class can define name, age, and course.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

Object

An object is an actual instance of a class.

```python
student1 = Student("Yangla", 25)
student2 = Student("Sonam", 24)
```

### Encapsulation

Encapsulation means keeping data and the methods that work on it together and controlling access to the data.

Example: Bank account.

You don't directly change your bank balance. You use methods such as deposit() and withdraw().

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance
```

### Inheritance

Inheritance means creating a new class from an existing class.

Real-world example:

Car → parent class
ElectricCar → child class

An electric car is a type of car, so it can inherit common car features.

```python
class Car:
    def start(self):
        print("Car starts")


class ElectricCar(Car):
    def charge(self):
        print("Battery is charging")


car = ElectricCar()

car.start()
car.charge()
```

### Types of Inheritance in Python

Inheritance means a child class can use the properties and methods of a parent class.

There are 5 main types:

1. Single Inheritance

One parent → One child

```python
class Animal:
    def eat(self):
        print("Animal eats")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

d = Dog()
d.eat()
d.bark()
```

2. Multiple Inheritance

Multiple parents → One child

```python
class Father:
    def skill1(self):
        print("Driving")

class Mother:
    def skill2(self):
        print("Cooking")

class Child(Father, Mother):
    pass

c = Child()
c.skill1()
c.skill2()
```

3. Multilevel Inheritance

Grandparent → Parent → Child

```python
class Grandfather:
    def house(self):
        print("Grandfather's house")

class Father(Grandfather):
    def car(self):
        print("Father's car")

class Son(Father):
    def bike(self):
        print("Son's bike")

s = Son()
s.house()
s.car()
s.bike()
```

4. Hierarchical Inheritance

One parent → Multiple children

```python
class Animal:
    def eat(self):
        print("Animal eats")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Cat(Animal):
    def meow(self):
        print("Cat meows")
```

5. Hybrid Inheritance

Combination of two or more types of inheritance.

For example, a combination of multiple + multilevel inheritance.

### Polymorphism

Polymorphism means the same method/action can behave differently for different objects.

Example: Animals make different sounds.

The method name is the same: sound(), but the behavior is different.

```python
class Dog:
    def sound(self):
        print("Dog barks")


class Cat:
    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```
